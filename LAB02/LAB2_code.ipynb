# Part3: Data Cleaning

## Before Clean

```python
print("Before Cleaning")
print("Rows:", df.shape[0])
print("Missing Year:", df['Year'].isnull().sum())
print("Missing Publisher:", df['Publisher'].isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

## Duplicate Removal

```python
df_cleaned = df.drop_duplicates().copy()
```

## Data Type Conversion + Incorrect Data Correction

```python
df_cleaned['Year'] = pd.to_numeric(df_cleaned['Year'], errors='coerce')
```

## Mean

```python
mean_year = df_cleaned['Year'].mean()

print("Mean:", mean_year)
```

## Median

```python
median_year = df_cleaned['Year'].median()

print("Median:", median_year)
```

## Missing Value Handling

```python
# เติม Year ด้วย median
df_cleaned['Year'] = df_cleaned['Year'].fillna(median_year)
df_cleaned['Year'] = df_cleaned['Year'].astype(int)

# เติม Publisher ด้วย mode
mode_pub = df_cleaned['Publisher'].mode()[0]
df_cleaned['Publisher'] = df_cleaned['Publisher'].fillna(mode_pub)
```

## After Clean

```python
print(df_cleaned.isnull().sum())
```

# Part4: Feature Engineering

## Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_cleaned['Genre_Label'] = le.fit_transform(df_cleaned['Genre'])

df_cleaned[['Genre', 'Genre_Label']].drop_duplicates().head()
```

## One-Hot Encoding

```python
df_onehot = pd.get_dummies(
    df_cleaned,
    columns=['Platform'],
    prefix='Platform',
    drop_first=True
)

df_onehot.head()
```
